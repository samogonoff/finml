# Исследование данных из Excel файла
## Загрузка и первичный анализ

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Путь к файлу (Windows)
file_path = r'D:\FinML\Реестр платежей март РФ.xlsx'

print(f"Файл: {file_path}")
print(f"Существует: {os.path.exists(file_path)}")
```

```python
# Загрузка данных
df_raw = pd.read_excel(file_path, header=None)
print(f"Размер: {df_raw.shape}")
print("\nПервые 10 строк:")
df_raw.head(10)
```

```python
# Поиск строки с заголовками
for i in range(10):
    print(f"Строка {i}: {df_raw.iloc[i].tolist()[:5]}...")
```

```python
# Загрузка с правильным смещением
df = pd.read_excel(file_path, header=5)
df.columns = df.columns.str.strip()

# Удаляем пустые колонки
cols_to_drop = [c for c in df.columns if 'Unnamed' in str(c)]
df = df.drop(columns=cols_to_drop, errors='ignore')

print(f"Размер после очистки: {df.shape}")
print(f"\nКолоники: {df.columns.tolist()}")
```

## Базовый анализ

```python
# Информация о данных
print("Типы данных:")
print(df.dtypes)

print("\nПропуски:")
print(df.isnull().sum())
```

```python
# Первые строки
df.head(10)
```

```python
# Статистика числовых колонок
df.describe()
```

## Анализ целевой переменной (Код УФ)

```python
target_col = 'Код УФ'
print(f"Целевая переменная: {target_col}")
print(f"Уникальных значений: {df[target_col].nunique()}")
print(f"Пропусков: {df[target_col].isnull().sum()} ({df[target_col].isnull().sum()/len(df)*100:.1f}%)")

print("\nТоп-15 кодов:")
df[target_col].value_counts().head(15)
```

```python
# Визуализация распределения кодов
plt.figure(figsize=(14, 6))
df[target_col].value_counts().head(20).plot(kind='bar')
plt.title('Топ-20 кодов УФ')
plt.xlabel('Код')
plt.ylabel('Количество')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Анализ текстовых полей

```python
# Анализ поля 'Информация'
print("Поле 'Информация':")
print(f"Пропусков: {df['Информация'].isnull().sum()}")
print(f"\nПримеры:")
for i, text in enumerate(df['Информация'].dropna().head(5)):
    print(f"\n{i+1}. {text[:200]}...")
```

```python
# Анализ поля 'Наименование'
print("Поле 'Наименование':")
print(f"Пропусков: {df['Наименование'].isnull().sum()}")
print(f"\nПримеры:")
for i, text in enumerate(df['Наименование'].dropna().head(10)):
    print(f"\n{i+1}. {text}")
```

## Извлечение Получателя из Наименования

```python
# Извлечение Получателя (символы до первого '/')
df['Получатель'] = df['Наименование'].str.split('/').str[0].str.strip()

print("Уникальных получателей:", df['Получатель'].nunique())
print("\nТоп-15 получателей:")
df['Получатель'].value_counts().head(15)
```

```python
# Анализ сумм
print("Сумма СНДС:")
print(df['Сумма СНДС'].describe())

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
df['Сумма СНДС'].hist(bins=50)
plt.title('Распределение сумм')
plt.xlabel('Сумма')

plt.subplot(1, 2, 2)
df['Сумма СНДС'].hist(bins=50, range=(0, 100000))
plt.title('Суммы до 100,000')
plt.xlabel('Сумма')

plt.tight_layout()
plt.show()
```

## Подготовка данных для модели

```python
# Создание текстового поля
df['Текст'] = df['Наименование'].fillna('').astype(str) + ' ' + df['Получатель'].fillna('')

# Данные с известным кодом
df_model = df.dropna(subset=['Код УФ']).copy()
print(f"Записей для модели: {len(df_model)} ({len(df_model)/len(df)*100:.1f}%)")

# Разделение кода
df_model['CFO'] = df_model['Код УФ'].astype(str).str[:-2]
df_model['PL'] = df_model['Код УФ'].astype(str).str[-2:]

print(f"Уникальных CFO: {df_model['CFO'].nunique()}")
print(f"Уникальных PL: {df_model['PL'].nunique()}")
```

```python
# Топ PL
print("Топ-10 PL:")
df_model['PL'].value_counts().head(10)
```

```python
# Топ CFO
print("Топ-10 CFO:")
df_model['CFO'].value_counts().head(10)
```

```python
# Сохранение подготовленных данных
df_model.to_pickle('/mnt/d/FinML/excel_data_prepared.pkl')
print("Данные сохранены в excel_data_prepared.pkl")
```
